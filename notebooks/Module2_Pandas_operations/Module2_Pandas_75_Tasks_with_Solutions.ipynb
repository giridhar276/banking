## Task 47: Create an amount category

## Task 48: Create a row-based review flag

Create a function that marks a transaction for Review when it is Failed or its Amount exceeds 20,000. Apply it row by row.

```python
def review_transaction(row):
    if row["Status"] == "Failed" or row["Amount"] > 20000:
        return "Review"
    return "Normal"

df["Review_Flag"] = df.apply(review_transaction, axis=1)
df[["Amount", "Status", "Review_Flag"]].head(10)
```

## Task 57: Use named aggregation

Create a summary by Account Type with Transaction Count, Total Amount, Average Amount and Maximum Amount.

```python
account_summary = df.groupby("Account_Type").agg(
    Transaction_Count=("Transaction_ID", "count"),
    Total_Amount=("Amount", "sum"),
    Average_Amount=("Amount", "mean"),
    Maximum_Amount=("Amount", "max")
).round(2)
account_summary
```

## Task 69: Matplotlib line chart

Create a simple line chart of Amount for the first ten transactions.

## Task 70: Matplotlib bar chart

## Task 72: Seaborn box plot

Create a box plot comparing Amount across Account Types.

## Task 73: Seaborn count plot

Create a count plot showing the number of transactions for each Status.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="Status", order=df["Status"].value_counts().index)
plt.title("Transaction Status Count")
plt.tight_layout()
plt.show()
```

## Task 74: Seaborn scatter plot

Create a scatter plot comparing Amount with Balance After Transaction and use Status as the color.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Amount",
    y="Balance_After_Transaction",
    hue="Status",
    s=80
)
plt.title("Amount vs Balance After Transaction")
plt.tight_layout()
plt.show()
```

## Task 75: Seaborn correlation heatmap

Create a heatmap for the numeric columns Amount and Balance After Transaction.

```python
numeric_columns = df[["Amount", "Balance_After_Transaction"]]
correlation = numeric_columns.corr()

plt.figure(figsize=(6, 4))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
