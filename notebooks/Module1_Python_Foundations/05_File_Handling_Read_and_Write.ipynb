# File Handling: Read and Write

## Learning objectives

- Open and close files safely
- Write, append and read text files
- Read and write a simple CSV file
- Use direct filenames without Path

This notebook uses small examples and explains each idea before the code.

## 1. File handling

Files allow information to remain available after a program finishes. In this notebook, we use direct filenames such as `customer.txt`. We do not use `Path`.

## 2. Writing a text file

Mode `w` creates a file or replaces its existing contents. The `with` statement closes the file automatically.

```python
with open("customer.txt", "w", encoding="utf-8") as file:
    file.write("Customer ID: 1001\n")
    file.write("Customer Name: Anita\n")
    file.write("Account Type: Savings\n")

print("customer.txt created")
```

## 3. Reading a text file

Mode `r` opens an existing file for reading.

```python
with open("customer.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

## 4. Appending information

Mode `a` adds new information at the end without deleting the existing contents.

```python
with open("customer.txt", "a", encoding="utf-8") as file:
    file.write("Account Status: Active\n")

with open("customer.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 5. Writing and reading CSV data

The built-in `csv` module helps us work with comma-separated data.

```python
import csv

transactions = [
    ["Transaction_ID", "Type", "Amount"],
    ["T001", "Deposit", 5000],
    ["T002", "Withdrawal", 1200]
]

with open("transactions.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(transactions)

with open("transactions.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Practice

Write three branch names to `branches.txt`, one per line, and then read the file.

```python
branches = ["Hyderabad", "Mumbai", "Chennai"]

with open("branches.txt", "w", encoding="utf-8") as file:
    for branch in branches:
        file.write(branch + "\n")

with open("branches.txt", "r", encoding="utf-8") as file:
    print(file.read())
```
