# Python Objects and Basic Data Types

## Learning objectives

- Understand that every value in Python is an object
- Work with numbers, strings, Boolean values, lists, tuples, sets and dictionaries

This notebook uses small examples and explains each idea before the code.

## 1. What is a Python object?

In Python, every value is an **object**. An object stores data and has a type. For example, `25` is an integer object and `"Savings"` is a string object.

```python
customer_name = "Anita"
age = 24
account_balance = 25000.50
is_active = True

print(customer_name)
print(type(customer_name))
print(type(age))
print(type(account_balance))
print(type(is_active))
```

## 2. Numbers and strings

Python commonly uses `int` for whole numbers, `float` for decimal numbers, and `str` for text.

```python
deposit = 5000
interest_rate = 6.5
account_type = "Savings Account"

print(deposit + 1000)
print(account_type.upper())
print(len(account_type))
```

## 3. Lists and tuples

A **list** is ordered and can be changed. A **tuple** is ordered but cannot be changed after creation.

```python
transactions = [1200, -500, 3000]
branch_codes = (101, 102, 103)

transactions.append(750)
print(transactions)
print(transactions[0])
print(branch_codes)
```

## 4. Sets and dictionaries

A **set** stores unique values. A **dictionary** stores information as key-value pairs.

```python
services = {"ATM", "UPI", "ATM", "Net Banking"}
customer = {
    "customer_id": 1001,
    "name": "Anita",
    "balance": 25000.50
}

print(services)
print(customer["name"])
customer["balance"] = 27000.50
print(customer)
```
