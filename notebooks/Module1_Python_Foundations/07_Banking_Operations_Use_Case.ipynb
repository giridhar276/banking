# Use Case: Simple Banking Operations

## Learning objectives

- Apply Python objects, conditions, loops and functions
- Perform deposit and withdrawal operations
- Maintain and display a transaction history
- Save a simple bank statement to a file

This notebook uses small examples and explains each idea before the code.

## Scenario

A small bank needs a simple program to manage one customer's account. The program should accept deposits, allow valid withdrawals, maintain transaction history and create a statement.

## 1. Create the account

A dictionary stores the account information, and a list stores transactions.

```python
account = {
    "account_number": "SB10001",
    "customer_name": "Anita",
    "balance": 10000.0
}

transactions = []

print(account)
```

## 2. Define banking functions

Each operation is placed in a separate function. This makes the program easier to read and reuse.

```python
def deposit(account, transactions, amount):
    if amount <= 0:
        print("Deposit amount must be positive")
        return

    account["balance"] = account["balance"] + amount
    transactions.append({"type": "Deposit", "amount": amount})
    print("Deposit successful")


def withdraw(account, transactions, amount):
    if amount <= 0:
        print("Withdrawal amount must be positive")
    elif amount > account["balance"]:
        print("Insufficient balance")
    else:
        account["balance"] = account["balance"] - amount
        transactions.append({"type": "Withdrawal", "amount": amount})
        print("Withdrawal successful")


def show_balance(account):
    print("Current balance: ₹", round(account["balance"], 2))
```

## 3. Perform sample operations

We use fixed values so freshers can run the notebook without entering input.

```python
deposit(account, transactions, 5000)
withdraw(account, transactions, 2500)
withdraw(account, transactions, 20000)
deposit(account, transactions, 1500)
show_balance(account)
```

## 4. Display transaction history

A loop displays every successful operation.

```python
print("Transaction History")
print("-------------------")

for number, transaction in enumerate(transactions, start=1):
    print(number, transaction["type"], "₹", transaction["amount"])
```

## 5. Save a bank statement

The statement is written using a direct filename. No `Path` object is used.

```python
with open("bank_statement.txt", "w", encoding="utf-8") as file:
    file.write("SIMPLE BANK STATEMENT\n")
    file.write("Account Number: " + account["account_number"] + "\n")
    file.write("Customer Name: " + account["customer_name"] + "\n")
    file.write("Transactions:\n")

    for transaction in transactions:
        line = transaction["type"] + ": " + str(transaction["amount"]) + "\n"
        file.write(line)

    file.write("Closing Balance: " + str(account["balance"]) + "\n")

print("Bank statement created successfully")
```

## 6. Read the generated statement

```python
with open("bank_statement.txt", "r", encoding="utf-8") as file:
    print(file.read())
```
