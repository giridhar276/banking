# Python Built-in Functions

## Learning objectives

- Use common built-in functions
- Convert values between basic data types
- Apply functions such as len, sum, min, max, sorted and round

This notebook uses small examples and explains each idea before the code.

## 1. What are built-in functions?

Python provides ready-to-use functions. We can call them directly without installing or importing anything.

```python
customer_name = "Meera"
balance = 15450.756

print(customer_name)
print(len(customer_name))
print(type(balance))
print(round(balance, 2))
```

## 2. Working with a list of numbers

The following functions quickly summarize numeric values.

```python
transactions = [1200, 4500, 800, 3200, 1500]

print("Total:", sum(transactions))
print("Smallest:", min(transactions))
print("Largest:", max(transactions))
print("Count:", len(transactions))
```

## 3. Sorting and range

`sorted()` returns values in order. `range()` creates a sequence of integers and is often used in loops.

```python
amounts = [4500, 1200, 3000, 900]
print(sorted(amounts))
print(sorted(amounts, reverse=True))

for number in range(1, 6):
    print("Token number:", number)
```

## 4. Type conversion

Data received from a user or a file may be text. Conversion functions change it into the required type.

```python
amount_text = "2500"
amount = int(amount_text)
rate = float("7.25")
reference_number = str(10025)

print(amount + 500)
print(rate)
print(reference_number, type(reference_number))
```

## 5. Enumerate and zip

`enumerate()` provides a position with each value. `zip()` combines corresponding items from multiple collections.

```python
customers = ["Asha", "Ravi", "John"]
balances = [10000, 15000, 9000]

for position, customer in enumerate(customers, start=1):
    print(position, customer)

for customer, balance in zip(customers, balances):
    print(customer, balance)
```

## Practice

Calculate the total, average, minimum and maximum of the given balances.

```python
balances = [12000, 18500, 9500, 22000]

total_balance = sum(balances)
average_balance = total_balance / len(balances)

print("Total:", total_balance)
print("Average:", round(average_balance, 2))
print("Minimum:", min(balances))
print("Maximum:", max(balances))
```
