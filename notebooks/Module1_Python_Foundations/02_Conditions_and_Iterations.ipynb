# Conditions and Iterations

## Learning objectives

- Make decisions using if, elif and else
- Repeat steps using for and while loops
- Use break and continue in simple situations

This notebook uses small examples and explains each idea before the code.

## 1. Conditions

Conditions allow a program to make decisions. Python uses indentation to show which statements belong to a condition.

```python
balance = 12000
withdrawal_amount = 5000

if withdrawal_amount <= balance:
    print("Withdrawal can be processed")
else:
    print("Insufficient balance")
```

## 2. Multiple conditions

Use `elif` when more than two outcomes are possible.

```python
credit_score = 735

if credit_score >= 750:
    print("Excellent credit profile")
elif credit_score >= 650:
    print("Acceptable credit profile")
else:
    print("Credit profile needs review")
```

## 3. For loop

A `for` loop processes items one by one.

```python
transactions = [1500, -300, 2200, -700]

for amount in transactions:
    if amount >= 0:
        print(amount, "Credit")
    else:
        print(amount, "Debit")
```

## 4. While loop

A `while` loop repeats as long as its condition remains true.

```python
installment_number = 1

while installment_number <= 5:
    print("Processing installment", installment_number)
    installment_number = installment_number + 1
```

## 5. Break and continue

`break` stops a loop. `continue` skips the current item and moves to the next one.

```python
amounts = [500, 0, 800, -1, 1200]

for amount in amounts:
    if amount == 0:
        continue
    if amount < 0:
        print("Invalid amount found. Stopping.")
        break
    print("Valid amount:", amount)
```
