# User-defined Functions

## Learning objectives

- Create functions with def
- Pass information through parameters
- Return results from functions
- Use default and keyword arguments

This notebook uses small examples and explains each idea before the code.

## 1. Why do we use functions?

A function groups related statements under one name. Functions reduce repeated code and make programs easier to understand and test.

```python
def greet_customer():
    print("Welcome to ABC Bank")

greet_customer()
```

## 2. Parameters and arguments

A parameter receives a value inside a function. An argument is the value supplied when the function is called.

```python
def greet_customer(name):
    print("Welcome", name)

greet_customer("Anita")
greet_customer("Rahul")
```

## 3. Returning a value

`return` sends a result back to the statement that called the function.

```python
def calculate_simple_interest(principal, rate, years):
    interest = principal * rate * years / 100
    return interest

result = calculate_simple_interest(50000, 7, 2)
print("Interest:", result)
```

## 4. Default and keyword arguments

A default value is used when an argument is not supplied. Keyword arguments make a function call easier to read.

```python
def calculate_fee(amount, fee_rate=1):
    return amount * fee_rate / 100

print(calculate_fee(5000))
print(calculate_fee(amount=5000, fee_rate=1.5))
```

## 5. Small functions working together

Breaking a problem into small functions is called functional decomposition.

```python
def is_valid_amount(amount):
    return amount > 0

def deposit(balance, amount):
    if is_valid_amount(amount):
        return balance + amount
    return balance

current_balance = 10000
current_balance = deposit(current_balance, 2500)
print("Updated balance:", current_balance)
```

## Practice

Complete a function that returns `True` when a customer has enough balance for a withdrawal.

```python
def can_withdraw(balance, amount):
    if amount > 0 and amount <= balance:
        return True
    return False

print(can_withdraw(10000, 3000))
print(can_withdraw(10000, 12000))
```
