# Classification metrics: a simple banking example

This notebook contains explanations only. **No Python code or model building is required.**

Imagine a bank checks **100 transactions** for fraud. We will use the same 100 transactions in every section. Here, **fraud is the positive class**.

1. Confusion matrix

**Simple definition:** A confusion matrix is a table that counts correct and incorrect predictions for each category.

A confusion matrix compares the **real answer** with the **model’s answer**.

| Actual transaction | Predicted fraud | Predicted genuine | Total |
|---|---:|---:|---:|
| **Fraud** | 8 | 2 | **10** |
| **Genuine** | 4 | 86 | **90** |
| **Total** | **12** | **88** | **100** |

Read the first row: of 10 real frauds, the model caught 8 and missed 2. Read the second row: of 90 genuine transactions, it wrongly flagged 4 and correctly cleared 86.

Check: 8 + 2 + 4 + 86 = 100.

2. TP: True Positive

**Simple definition:** A true positive is a real fraud transaction that the model correctly calls fraud.

**8 fraud transactions were correctly flagged as fraud.**

**True** means the prediction was correct. **Positive** means fraud in this example.

Think: **Fraud caught = 8**.

3. FN: False Negative

**Simple definition:** A false negative is a real fraud transaction that the model incorrectly calls genuine.

**2 fraud transactions were predicted as genuine.** These frauds were missed.

**False** means the prediction was wrong. **Negative** means the model said genuine.

Think: **Fraud missed = 2**.

4. FP: False Positive

**Simple definition:** A false positive is a genuine transaction that the model incorrectly calls fraud.

**4 genuine transactions were predicted as fraud.** These are false alarms.

**False** means the prediction was wrong. **Positive** means the model said fraud.

Think: **Genuine transactions wrongly blocked or reviewed = 4**.

5. TN: True Negative

**Simple definition:** A true negative is a genuine transaction that the model correctly calls genuine.

**86 genuine transactions were correctly predicted as genuine.**

**True** means the prediction was correct. **Negative** means the model said genuine.

Think: **Genuine transactions correctly cleared = 86**.

6. Precision

**Simple definition:** Precision tells us what share of fraud alerts are actually fraud.

**Question: When the model raises a fraud alert, how often is that alert right?**

The model raised **12 alerts**: 8 real frauds + 4 false alarms.

**Precision = TP / (TP + FP) = 8 / (8 + 4) = 8/12 ≈ 67%.**

Meaning: out of every 12 alerts in this example, **8 really were fraud**. If you want fewer false alarms, pay attention to precision.

**Memory aid: Start with what the model flagged.**

7. Recall

**Simple definition:** Recall tells us what share of all real fraud transactions the model catches.

**Question: Of all real frauds, how many did the model catch?**

There were **10 real frauds**: 8 caught + 2 missed.

**Recall = TP / (TP + FN) = 8 / (8 + 2) = 8/10 = 80%.**

Meaning: the model caught **8 of the 10 actual frauds**. If missing fraud is costly, pay attention to recall.

**Memory aid: Start with what actually was fraud.**

8. F1 score

**Simple definition:** F1 is one score that combines precision and recall. It is high when both are high.

**Question: What single score summarizes both precision and recall?**

**F1 = 2 × (precision × recall) / (precision + recall).** Here precision is 8/12 and recall is 8/10, so **F1 ≈ 73%**. You can also use **F1 = 2TP / (2TP + FP + FN) = 16 / 22 ≈ 73%**.

F1 becomes lower if either precision or recall is low. It is helpful when you want to consider **both false alarms and missed fraud**.

**Important:** 73% is a combined score; it does **not** mean 73% of all transactions were correct.

9. Sensitivity

**Simple definition:** Sensitivity tells us what share of real fraud transactions the model catches. It is exactly the same measure as recall.

**Sensitivity is another name for recall.**

**Question: Of all actual frauds, how many did we catch?**

**Sensitivity = TP / (TP + FN) = 8/10 = 80%.**

If a report says *sensitivity 80%* and another says *recall 80%*, they mean the same thing here. It is also called the **true positive rate (TPR)**.

10. Specificity

**Simple definition:** Specificity tells us what share of genuine transactions the model correctly identifies as genuine.

**Question: Of all actual genuine transactions, how many did the model correctly clear?**

There were **90 genuine transactions**: 86 cleared + 4 wrongly flagged.

**Specificity = TN / (TN + FP) = 86 / (86 + 4) = 86/90 ≈ 96%.**

Meaning: the model correctly cleared **86 of 90 genuine transactions**. Specificity is also called the **true negative rate (TNR)**.

**Memory aid: Start with what actually was genuine.**

11. Accuracy

**Simple definition:** Accuracy tells us what share of all predictions, fraud and genuine combined, are correct.

**Question: Out of all transactions, how many predictions were correct?**

Correct predictions = 8 frauds caught + 86 genuine transactions cleared = **94**.

**Accuracy = (TP + TN) / total = (8 + 86) / 100 = 94%.**

Accuracy sounds strong, but notice that **2 of 10 frauds were missed**. Since most transactions are genuine, accuracy alone can hide poor fraud detection. A model that calls all 100 transactions genuine would score **90% accuracy but 0% fraud recall**.

12. ROC curve and AUC

**Simple definition (ROC curve):** An ROC curve shows how the share of frauds caught and the share of genuine transactions falsely flagged change as we adjust the fraud alert cutoff.

**Simple definition (AUC):** AUC is a single number that summarizes the ROC curve across all cutoffs; higher means the model generally separates fraud from genuine transactions better.

Many models first give a **fraud score** for each transaction. A cutoff decides when a score becomes a fraud alert. For example, a bank might flag scores above 0.50.

Lowering the cutoff usually catches more fraud **and** raises more false alarms. Raising it usually does the opposite.

An **ROC curve** shows that tradeoff across many cutoffs:

- **Vertical axis:** Recall / Sensitivity / True Positive Rate = TP / (TP + FN).
- **Horizontal axis:** False Positive Rate = FP / (FP + TN) = 1 − Specificity.

For the confusion matrix above, the point at its chosen cutoff is **(4/90 ≈ 4.4% false positive rate, 8/10 = 80% recall)**. One confusion matrix gives **one point**, not a full ROC curve.

**AUC** is the area under the full ROC curve. It ranges from 0 to 1: **0.5** is roughly random ranking; **1.0** is perfect ranking. To calculate ROC–AUC you need scores at different cutoffs, not just this one confusion matrix. We therefore cannot calculate a numeric AUC from the four counts above.

13. ROC–AUC versus accuracy

**Simple distinction:** Accuracy checks predictions at one chosen cutoff. ROC–AUC summarizes how the model separates the two classes across many cutoffs.

**Accuracy** asks: *How many labels were correct at the chosen cutoff?* Our answer is **94%**.

**ROC–AUC** asks: *Across different cutoffs, how well does the model tend to place fraud cases ahead of genuine cases?* We cannot give its value without fraud scores.

ROC–AUC does not tell you how many frauds will be caught at the cutoff the bank actually uses. For that decision, inspect recall, false alarms, precision, and the confusion matrix **at that cutoff**.

14. Quick recap

| Measure | Start with | Plain question | This example |
|---|---|---|---:|
| Precision | 12 predicted fraud alerts | How many alerts were real fraud? | 8/12 ≈ 67% |
| Recall / Sensitivity | 10 actual frauds | How many frauds were caught? | 8/10 = 80% |
| Specificity | 90 actual genuine transactions | How many genuine transactions were cleared? | 86/90 ≈ 96% |
| F1 | Precision and recall | How well do those two scores work together? | ≈ 73% |
| Accuracy | All 100 transactions | How many predictions were correct? | 94/100 = 94% |
| ROC–AUC | Fraud scores across cutoffs | How well does the model rank fraud above genuine? | Cannot calculate from this table |

**One sentence to remember:** Precision checks **alerts**; recall checks **real frauds**; specificity checks **real genuine transactions**.
