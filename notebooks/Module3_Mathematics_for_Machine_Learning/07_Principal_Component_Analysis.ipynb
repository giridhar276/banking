# Principal Component Analysis

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Understand why dimensionality reduction is useful
- Standardize features before PCA
- Reduce three banking features to two components

## 1. What is PCA?

A dataset may contain many related columns. Principal Component Analysis, or PCA, combines them into fewer new columns called principal components. The goal is to keep as much useful variation as possible.

```python
import pandas as pd

data = pd.DataFrame({
    "Income": [40000, 50000, 65000, 80000, 95000, 55000],
    "Loan_Amount": [120000, 150000, 180000, 250000, 300000, 160000],
    "Credit_Score": [650, 680, 710, 740, 780, 695]
})

data
```

## 2. Standardize the values

The columns use different units. Income is in thousands while credit score is in hundreds. Standardization puts them on a comparable scale.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
```

## 3. Apply PCA

We reduce three original columns to two principal components.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
components = pca.fit_transform(scaled_data)

pca_data = pd.DataFrame(components, columns=["PC1", "PC2"])
pca_data
```

## 4. Explained variance

Explained variance tells us how much information each component keeps. Values closer to 1 mean more information is retained.

```python
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total information retained:", round(pca.explained_variance_ratio_.sum() * 100, 2), "%")
```

## 5. Visualize the components

Each point represents one customer in the new two-dimensional space.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))
plt.scatter(pca_data["PC1"], pca_data["PC2"], color="teal", s=80)
plt.title("Customers After PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Important note

Principal components are combinations of the original columns. They are useful for reducing dimensions, but they may be harder to explain directly than Income or Credit Score.
