# Singular Value Decomposition

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Understand SVD in simple terms
- Break a matrix into three smaller parts
- Create a simple lower-detail reconstruction

## 1. What is SVD?

Singular Value Decomposition, or SVD, breaks one matrix into three parts: `U`, `S` and `Vᵀ`. The largest singular values represent the strongest patterns in the data.

```python
import numpy as np

data = np.array([
    [5, 4, 0],
    [4, 5, 1],
    [1, 0, 5],
    [0, 1, 4]
], dtype=float)

print(data)
```

## 2. Apply SVD

NumPy returns `U`, the singular values, and `Vᵀ`.

```python
U, singular_values, VT = np.linalg.svd(data, full_matrices=False)

print("U shape:", U.shape)
print("Singular values:", np.round(singular_values, 2))
print("V transpose shape:", VT.shape)
```

## 3. Reconstruct the original matrix

When all singular values are used, we can rebuild the original matrix.

```python
S = np.diag(singular_values)
reconstructed = U @ S @ VT

print(np.round(reconstructed, 2))
```

## 4. Keep only the strongest pattern

Using only the first singular value produces a simpler version of the data. This is called a rank-1 approximation.

```python
U_small = U[:, :1]
S_small = np.diag(singular_values[:1])
VT_small = VT[:1, :]

rank_one = U_small @ S_small @ VT_small
print(np.round(rank_one, 2))
```

## 5. Compare the two matrices

The simplified matrix is not identical, but it keeps the strongest overall pattern.

```python
print("Original:\n", data)
print("Simplified:\n", np.round(rank_one, 2))
print("Average error:", round(np.mean(np.abs(data - rank_one)), 2))
```

## Banking connection

SVD can help compress large tables, discover patterns and support recommendation systems. For freshers, the important idea is that it keeps the strongest information using fewer components.
