# Matrix Operations

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Add, subtract and multiply matrices
- Understand transpose and dot product
- Create identity and inverse matrices

## 1. Create two small matrices

We use small values so every result is easy to inspect.

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A:\n", A)
print("B:\n", B)
```

## 2. Addition and subtraction

Matrices of the same shape can be added or subtracted position by position.

```python
print("Addition:\n", A + B)
print("Subtraction:\n", B - A)
```

## 3. Scalar multiplication

Multiplying a matrix by one number changes every value in the matrix.

```python
print(A * 3)
```

## 4. Matrix multiplication

The `@` operator performs matrix multiplication. It combines rows from the first matrix with columns from the second.

```python
result = A @ B
print(result)
```

## 5. Transpose

Transpose changes rows into columns and columns into rows.

```python
print("Original:\n", A)
print("Transpose:\n", A.T)
```

## 6. Dot product

A dot product multiplies matching vector values and adds the results. ML models use it to combine feature values with weights.

```python
features = np.array([50000, 200000, 720])
weights = np.array([0.00001, 0.000002, 0.01])
score = np.dot(features, weights)
print("Combined score:", round(score, 2))
```

## 7. Identity and inverse

An identity matrix behaves like the number 1 in matrix multiplication. An inverse can reverse the effect of a matrix when an inverse exists.

```python
identity = np.eye(2)
inverse_A = np.linalg.inv(A)

print("Identity:\n", identity)
print("Inverse of A:\n", inverse_A)
print("A multiplied by its inverse:\n", np.round(A @ inverse_A, 2))
```
