# Derivatives and Gradients

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Understand slope and derivative
- Estimate a derivative using Python
- Understand a gradient with two variables

## 1. Function and slope

A function converts an input into an output. A derivative tells us how quickly the output changes when the input changes.

```python
def square(x):
    return x ** 2

print(square(2))
print(square(3))
```

## 2. Estimate a derivative

We can estimate the slope by making a very small change to `x` and measuring the output change.

```python
def numerical_derivative(function, x, small_change=0.0001):
    output_change = function(x + small_change) - function(x)
    return output_change / small_change

print("Estimated derivative at x=3:", numerical_derivative(square, 3))
```

## 3. Compare with the known derivative

For `x²`, the derivative is `2x`. At `x=3`, the exact derivative is therefore 6.

```python
x = 3
exact_derivative = 2 * x
print("Exact derivative:", exact_derivative)
```

## 4. Partial derivatives

A function may have more than one input. A partial derivative measures change with respect to one input while keeping the other input fixed.

```python
def loss(x, y):
    return x ** 2 + y ** 2

x = 2
y = 3
partial_x = 2 * x
partial_y = 2 * y

print("Loss:", loss(x, y))
print("Partial derivative for x:", partial_x)
print("Partial derivative for y:", partial_y)
```

## 5. Gradient

A gradient collects all partial derivatives into one vector. It points in the direction where the function increases fastest.

```python
import numpy as np

gradient = np.array([partial_x, partial_y])
print("Gradient:", gradient)
```

## Machine-learning connection

A model has a loss value that measures error. Gradients tell the training process how to change model parameters to reduce that error.
