# Banking Credit-Risk Mathematics Use Case

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Apply vectors, matrices and weighted scores
- Standardize banking features
- Use PCA to visualize customer patterns

## Scenario

A bank wants to explore customer credit information. This is a learning example, not a production credit-decision system. We will use simple mathematics to organize and visualize the data.

```python
import pandas as pd
import numpy as np

customers = pd.DataFrame({
    "Customer_ID": [f"CUST{i:03d}" for i in range(1, 13)],
    "Income": [42000, 58000, 76000, 36000, 91000, 67000, 48000, 82000, 52000, 105000, 44000, 73000],
    "Loan_Amount": [150000, 190000, 240000, 130000, 280000, 210000, 170000, 260000, 180000, 320000, 145000, 230000],
    "Credit_Score": [640, 690, 735, 610, 780, 710, 665, 750, 680, 805, 625, 725],
    "Late_Payments": [4, 2, 1, 6, 0, 1, 3, 0, 2, 0, 5, 1],
    "Risk_Group": ["High", "Medium", "Low", "High", "Low", "Low", "Medium", "Low", "Medium", "Low", "High", "Low"]
})

customers
```

## 1. Represent one customer as a vector

The vector contains Income, Loan Amount, Credit Score and Late Payments for the first customer.

```python
first_customer_vector = customers.loc[0, ["Income", "Loan_Amount", "Credit_Score", "Late_Payments"]].to_numpy(dtype=float)
print(first_customer_vector)
```

## 2. Represent all customers as a matrix

Rows represent customers and columns represent numeric features.

```python
feature_names = ["Income", "Loan_Amount", "Credit_Score", "Late_Payments"]
feature_matrix = customers[feature_names].to_numpy(dtype=float)

print(feature_matrix)
print("Matrix shape:", feature_matrix.shape)
```

## 3. Calculate a simple learning score

A dot product combines standardized feature values with example weights. This score is only for understanding matrix mathematics; it is not a real approval score.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(feature_matrix)
weights = np.array([-0.20, 0.25, -0.35, 0.40])
learning_scores = scaled_features @ weights

customers["Learning_Risk_Score"] = np.round(learning_scores, 2)
customers[["Customer_ID", "Learning_Risk_Score", "Risk_Group"]]
```

## 4. Reduce four features to two components

PCA creates two new columns that retain the strongest patterns across the four original features.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
components = pca.fit_transform(scaled_features)

customers["PC1"] = components[:, 0]
customers["PC2"] = components[:, 1]

print("Information retained:", round(pca.explained_variance_ratio_.sum() * 100, 2), "%")
customers[["Customer_ID", "PC1", "PC2", "Risk_Group"]]
```

## 5. Visualize customer groups

The chart helps us see whether customers in the same risk group appear near one another after PCA.

```python
import matplotlib.pyplot as plt

colors = {"Low": "green", "Medium": "orange", "High": "red"}

plt.figure(figsize=(8, 5))
for risk_group in ["Low", "Medium", "High"]:
    group = customers[customers["Risk_Group"] == risk_group]
    plt.scatter(
        group["PC1"],
        group["PC2"],
        label=risk_group,
        color=colors[risk_group],
        s=90
    )

plt.title("Credit-Risk Groups After PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Risk Group")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Interpretation

- A vector represents one customer's numeric features.
- A matrix represents all customers.
- A dot product combines features with weights.
- Standardization makes different measurement units comparable.
- PCA reduces the number of dimensions.
- Human review, fairness testing and proper model validation would be required before using any credit-risk system in practice.
