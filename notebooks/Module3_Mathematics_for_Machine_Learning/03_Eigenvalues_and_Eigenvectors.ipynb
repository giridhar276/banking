# Eigenvalues and Eigenvectors

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Understand the basic idea of eigenvectors
- Calculate eigenvalues and eigenvectors with NumPy
- Verify the result using a simple calculation

## 1. Simple meaning

A matrix can change the length and direction of a vector. An **eigenvector** is a special direction that does not turn after the matrix transformation. Its length may change. The amount of change is the **eigenvalue**.

```python
import numpy as np

matrix = np.array([[2, 0], [0, 3]])
print(matrix)
```

## 2. Calculate eigenvalues and eigenvectors

NumPy performs the calculation for us using `np.linalg.eig()`.

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

## 3. Read the result

Each eigenvector is stored as a column. The first eigenvalue belongs to the first eigenvector.

```python
first_value = eigenvalues[0]
first_vector = eigenvectors[:, 0]

print("First eigenvalue:", first_value)
print("First eigenvector:", first_vector)
```

## 4. Verify the relationship

For an eigenvector, multiplying the matrix by the vector gives the same result as multiplying the vector by its eigenvalue.

```python
left_side = matrix @ first_vector
right_side = first_value * first_vector

print("Matrix × vector:", left_side)
print("Eigenvalue × vector:", right_side)
print("Are they equal?", np.allclose(left_side, right_side))
```

## 5. Why does ML use this?

Principal Component Analysis uses eigenvectors to find important directions in data. These directions help reduce the number of columns while keeping useful information.

## Practice with another matrix

```python
banking_matrix = np.array([[4, 2], [1, 3]])
values, vectors = np.linalg.eig(banking_matrix)

print("Eigenvalues:", np.round(values, 2))
print("Eigenvectors:\n", np.round(vectors, 2))
```
