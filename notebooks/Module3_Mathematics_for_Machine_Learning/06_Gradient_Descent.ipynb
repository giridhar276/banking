# Gradient Descent

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Understand how gradient descent reduces error
- Use a learning rate to update a value
- Visualize the path toward a minimum

## 1. Simple goal

We want to find the value of `x` that makes `(x - 3)²` as small as possible. The best answer is `x = 3`, where the loss becomes zero.

```python
def loss(x):
    return (x - 3) ** 2

def gradient(x):
    return 2 * (x - 3)

print("Loss at x=8:", loss(8))
```

## 2. Update rule

Gradient descent repeatedly uses: `new value = old value - learning rate × gradient`. The learning rate controls the size of each step.

```python
x = 8.0
learning_rate = 0.1

new_x = x - learning_rate * gradient(x)
print("Old x:", x)
print("New x:", new_x)
print("New loss:", loss(new_x))
```

## 3. Repeat the updates

Each update moves `x` closer to 3 and reduces the loss.

```python
x = 8.0
learning_rate = 0.1
history = []

for step in range(20):
    history.append((step, x, loss(x)))
    x = x - learning_rate * gradient(x)

for step, value, current_loss in history[:5]:
    print(step, round(value, 4), round(current_loss, 4))

print("Final x:", round(x, 4))
print("Final loss:", round(loss(x), 6))
```

## 4. Visualize the loss

The chart should move downward because the error becomes smaller.

```python
import matplotlib.pyplot as plt

steps = [item[0] for item in history]
losses = [item[2] for item in history]

plt.figure(figsize=(7, 4))
plt.plot(steps, losses, marker="o")
plt.title("Loss During Gradient Descent")
plt.xlabel("Step")
plt.ylabel("Loss")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 5. Two-variable example

The same idea works with more than one parameter. Here both `x` and `y` move toward zero.

```python
x = 5.0
y = -4.0
learning_rate = 0.1

for step in range(20):
    gradient_x = 2 * x
    gradient_y = 2 * y
    x = x - learning_rate * gradient_x
    y = y - learning_rate * gradient_y

print("Final x:", round(x, 4))
print("Final y:", round(y, 4))
```
