# Scalars, Vectors and Matrices

This notebook explains the mathematics in simple terms. You do not need advanced mathematics to follow the examples.

## Learning objectives

- Understand scalars, vectors and matrices
- Create them using NumPy
- Check their shape and access individual values

## 1. Why do we need these objects?

Machine-learning data is stored as numbers. A single number is a **scalar**, a row of numbers is a **vector**, and a table of numbers is a **matrix**.

```python
import numpy as np
```

## 2. Scalar

A scalar is one value. A customer's account balance is a simple example.

```python
account_balance = 25000
print("Scalar value:", account_balance)
```

## 3. Vector

A vector is a one-dimensional collection of numbers. It can represent several features of one customer.

```python
customer_features = np.array([35, 60000, 720])
print("Vector:", customer_features)
print("Shape:", customer_features.shape)
```

## 4. Matrix

A matrix is a table of numbers. Rows can represent customers and columns can represent features.

```python
customers = np.array([
    [35, 60000, 720],
    [42, 85000, 690],
    [29, 45000, 760]
])

print(customers)
print("Shape:", customers.shape)
```

## 5. Accessing values

NumPy starts counting at zero. The first position is therefore index 0.

```python
print("First customer:", customers[0])
print("First customer's income:", customers[0, 1])
print("All credit scores:", customers[:, 2])
```
