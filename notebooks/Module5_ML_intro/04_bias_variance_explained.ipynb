## 9. Bias — when the model makes assumptions that are too simple

**Explanation** Think of bias as how far the model’s predictions are consistently from the correct answer.

Suppose you train a model to predict a customer’s monthly income. You train it once using 100 customers, then again using a different 100 customers. For a new customer whose actual income is ₹80,000:

Model 1 predicts ₹50,000.
Model 2 predicts ₹52,000.

The predictions are similar, but both are far too low. That pattern suggests high bias: the model keeps making the same kind of mistake, perhaps because it uses too few relevant details.

Easy distinction: Bias is a consistent mistake. Variance is predictions changing a lot when the training data changes.

**---------------**

**Bias** is the tendency to make a consistent kind of prediction error because the model misses part of the real relationship.

**Example** Bias is error caused by a model making overly simple assumptions. For example, a bank’s loan model might predict income using only years of experience, ignoring job role and location. It may make similar mistakes across many records.

**Banking example:** This notebook predicts `Balance_After_Transaction` using only `Amount`. Suppose two customers each spend ₹1,000, but one started with ₹2,000 and the other with ₹50,000. Their final balances differ greatly. A model using only the amount cannot tell them apart, regardless of how carefully it learns the straight line.

**What to look for:** If both training and test R² are poor, missing useful inputs can be one reason for high bias. Consider adding the balance before the transaction, if the dataset has that column. Bias is about the model's assumptions; one low score by itself does not measure bias directly.


## 10. Variance — when predictions depend too much on the training sample

**Explanation:** Think of variance as how much the model changes when the training data changes.

Suppose you train a loan approval model using 100 customers. Then you train the same model again using a different 100 customers:

If both models give similar predictions for the same new applicant, variance is low.
If one says “approve” and the other says “reject,” variance is high.

A high-variance model is too sensitive to the particular customers it learned from. It may do very well on those customers but struggle with new ones.

**------**

**Variance** describes how much a model's learned predictions change if it is trained on a different set of records.

**Example** Variance is how much a model’s predictions change when it learns from a different sample of training data. A model with high variance may learn small details of one sample, then perform poorly on new applicants.

**Banking example:** Imagine a flexible model trained on a few transactions. It might learn a rule that works for one unusual ₹25,000 transaction but changes substantially when that record is replaced. Such a model has high variance.

**What to look for:** Training R² may be high while test R² is much lower. That gap can suggest high variance, especially with a complex model. This notebook's straight-line model is relatively simple; comparing its two scores is a useful first check, but a small test set makes the gap unstable.


## 11. Underfitting — when the model learns too little

**Underfitting** occurs when a model cannot capture the important pattern, even for the data used to train it. It is commonly associated with high bias.

**Banking example:** If starting balance strongly determines final balance, a straight-line model that sees only the transaction amount will fail to explain many balances.

**How to recognize it:** Training R² is low **and** test R² is also low. Look at the two R² values printed above. If both are poor, underfitting is one possible explanation; check which inputs are available and inspect the prediction errors before deciding.


## 12. Overfitting — when the model learns the training records too closely

**Overfitting** occurs when a model learns details or random noise in its training records that do not repeat in new records. It is commonly associated with high variance.

**Banking example:** A very complex model might memorize the final balances of the training transactions. It predicts those records well but struggles with a new customer transaction.

**How to recognize it:** Training R² is high, while test R² is much lower. Compare the two scores above. A gap is a clue rather than proof, especially when this notebook has only a small test set. More unseen records and a simpler model can help check whether the pattern is real.


**Final** high bias often leads to underfitting, while high variance often leads to overfitting.
