# ResNet50 — Identify the main content of an image

Use case: automatically suggest a category for a photo in a digital image library. ResNet50 is a CNN with shortcut connections. We use weights already learned on ImageNet; no training is needed.

## Step 2 — Import libraries
TensorFlow runs the neural network. Keras provides the model API. NumPy handles arrays and Matplotlib displays pictures.


```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras

keras.utils.set_random_seed(42)
print("TensorFlow:", tf.__version__)
```

## Step 3 — Load and display a picture
Pixels are the numeric representation of an image. Display the original before making predictions.


```python
image_path = keras.utils.get_file(
    "sunflower.jpg",
    "https://storage.googleapis.com/download.tensorflow.org/example_images/592px-Red_sunflower.jpg"
)
# To use your own picture, replace image_path with a local filename.
original = keras.utils.load_img(image_path)
plt.figure(figsize=(6, 6))
plt.imshow(original)
plt.axis("off")
plt.show()
```

## Step 4 — Load ResNet50
weights="imagenet" loads learned weights. The original classifier predicts 1,000 ImageNet categories. These are not your own custom categories.


```python
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
model = ResNet50(weights="imagenet")
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
```

## Step 5 — Prepare the image
Resize to 224 × 224. Add a batch dimension because the model expects a group of images. ResNet preprocessing uses its training-time channel order and centering; do not replace it with division by 255.


```python
picture = keras.utils.load_img(image_path, target_size=(224, 224))
pixels = keras.utils.img_to_array(picture)
batch = np.expand_dims(pixels, axis=0)
model_input = preprocess_input(batch.copy())
print("Batch shape:", model_input.shape)
```

## Step 6 — Predict and display the top categories
A score measures the model’s preference for a category. It is not a guarantee of correctness.


```python
predictions = model.predict(model_input, verbose=0)
top_predictions = decode_predictions(predictions, top=5)[0]
for category_id, label, score in top_predictions:
    print(f"{label}: {score:.2%}")
plt.imshow(picture)
plt.title(f"Prediction: {top_predictions[0][1]} ({top_predictions[0][2]:.1%})")
plt.axis("off")
plt.show()
```

## Understand the result
This is image classification: one category for the overall picture. It does not locate objects or create bounding boxes. Unfamiliar objects can still receive a confident but incorrect label.
